```python
# Run first — confirms the notebook uses the project virtual environment
import sys

print("Python:", sys.executable)

if ".venv" not in sys.executable:
    raise RuntimeError(
        "Wrong kernel. Select: .venv/bin/python\n"
        "Kernel picker (top-right) → Python Environments → .venv"
    )

import transformers
import bertviz
print("transformers:", transformers.__version__)
print("Ready.")
```

```python
from transformers import AutoTokenizer, AutoModel, utils
from bertviz import model_view, head_view

utils.logging.set_verbosity_error()

model_name = "bert-base-uncased"
sentence = "The cat sat on the mat because it was tired."
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name, output_attentions=True)

inputs = tokenizer(sentence, return_tensors="pt")
outputs = model(**inputs)
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

print("Tokens:")
print(tokens)
print("\nNumber of layers:", len(outputs.attentions))
print("Attention shape (layer 0):", tuple(outputs.attentions[0].shape))
```

## Model View
Bird's-eye view of attention across **all layers and heads**.

```python
model_view(outputs.attentions, tokens)
```

## Head View
Click tokens — especially **"it"** — and compare attention lines across heads.

```python
head_view(outputs.attentions, tokens)
```
